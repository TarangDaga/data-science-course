```python


import numpy as np
import time

print("===================================")
print("🎲 WELCOME TO DICE ROLLER GAME 🎲")
print("===================================")

# Roll history store karne ke liye
history = []

# Main game loop
while True:

    print("\n--------- MENU ---------")
    print("1. Roll Dice")
    print("2. View History")
    print("3. Game Statistics")
    print("4. Exit")

    choice = input("\nEnter your choice: ")

    # ==============================
    # OPTION 1 → ROLL DICE
    # ==============================
    if choice == "1":

        # User Input
        dice_count = int(input("\nKitne dice roll karne hai? : "))
        dice_sides = int(input("Dice me kitni sides hai? : "))

        print("\n🎲 Rolling Dice...", end=" ")

        # Small animation
        for i in range(3):
            print(".", end=" ")
            time.sleep(0.5)

        print("\n")

        # NumPy se random dice generate
        rolls = np.random.randint(1, dice_sides + 1, size=dice_count)

        # Total calculate
        total = np.sum(rolls)

        # Result show
        print("✨ Dice Results :", rolls)
        print("🔥 Total Score  :", total)

        # Special Conditions
        if np.all(rolls == rolls[0]):
            print("🎉 Jackpot! Sare dice same aaye!")

        elif total > (dice_count * dice_sides) * 0.7:
            print("💥 High Score Roll!")

        elif total < (dice_count * dice_sides) * 0.3:
            print("😢 Low Score Roll!")

        # History save
        history.append(total)

    # ==============================
    # OPTION 2 → HISTORY
    # ==============================
    elif choice == "2":

        print("\n📜 Roll History")

        if len(history) == 0:
            print("No history available!")

        else:
            for i, score in enumerate(history, start=1):
                print(f"Roll {i} : {score}")

    # ==============================
    # OPTION 3 → STATISTICS
    # ==============================
    elif choice == "3":

        print("\n📊 GAME STATISTICS")

        if len(history) == 0:
            print("No data available!")

        else:
            print("Total Games Played :", len(history))
            print("Highest Score      :", np.max(history))
            print("Lowest Score       :", np.min(history))
            print("Average Score      :", round(np.mean(history), 2))

    # ==============================
    # OPTION 4 → EXIT
    # ==============================
    elif choice == "4":

        print("\n👋 Thanks for playing Dice Roller Game!")
        break

    # ==============================
    # INVALID INPUT
    # ==============================
    else:
        print("\n❌ Invalid Choice! Try Again.")
```
